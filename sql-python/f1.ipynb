```python
# Creamos la base de datos y cargamos los archivos CSV en tablas
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

#Created on Thu Oct 19 11:00:00 2023

import pandas as pd
from sqlalchemy import create_engine, text

# Define la ruta donde tienes los archivos CSV
data_path = './dataset_f1/'  # Asegúrate de que los archivos CSV estén en este directorio o ajusta la ruta

# Define el nombre de la base de datos SQLite
db_name = 'formula1.sqlite'
engine = create_engine(f'sqlite:///{db_name}')

# Lista de archivos CSV y sus correspondientes nombres de tabla
csv_files = {
    'circuits.csv': 'circuits',
    'constructor_results.csv': 'constructor_results',
    'constructor_standings.csv': 'constructor_standings',
    'constructors.csv': 'constructors',
    'driver_standings.csv': 'driver_standings',
    'drivers.csv': 'drivers',
    'lap_times.csv': 'lap_times',
    'pit_stops.csv': 'pit_stops',
    'qualifying.csv': 'qualifying',
    'races.csv': 'races',
    'results.csv': 'results',
    'seasons.csv': 'seasons',
    'sprint_results.csv': 'sprint_results',
    'status.csv': 'status'
}

try:
    # Itera sobre los archivos CSV y los carga en la base de datos
    for file_name, table_name in csv_files.items():
        file_path = f'{data_path}{file_name}'
        try:
            df = pd.read_csv(file_path)
            df.to_sql(table_name, engine, if_exists='replace', index=False)
            print(f'Tabla "{table_name}" cargada exitosamente desde "{file_name}".')
        except FileNotFoundError:
            print(f'Error: El archivo "{file_path}" no fue encontrado.')
        except Exception as e:
            print(f'Error al cargar la tabla "{table_name}" desde "{file_name}": {e}')

    print('\n¡Todas las tablas han sido procesadas!')

except Exception as e:
    print(f'Ocurrió un error general: {e}')
finally:
    # Cierra la conexión a la base de datos
    engine.dispose()
    print('\nConexión a la base de datos cerrada.')
```

```python
from sqlalchemy import create_engine, inspect

# Define el nombre de la base de datos SQLite
db_name = 'formula1.sqlite'
engine = create_engine(f'sqlite:///{db_name}')

# Crea un objeto inspector para examinar la base de datos
inspector = inspect(engine)

# Obtiene la lista de todos los nombres de las tablas
table_names = inspector.get_table_names()

print('Esquema de la base de datos:')
for table_name in table_names:
    print(f'\nTabla: {table_name}')
    columns = inspector.get_columns(table_name)
    print('  Columnas:')
    for column in columns:
        print(f'    - {column["name"]} ({column["type"]}), Primary Key: {column["primary_key"]}, Nullable: {column["nullable"]}')
    foreign_keys = inspector.get_foreign_keys(table_name)
    if foreign_keys:
        print('  Claves Foráneas:')
        for fk in foreign_keys:
            referenced_table = fk["referred_table"]
            local_columns = fk["constrained_columns"]
            referenced_columns = fk["referred_columns"]
            print(f'    - {local_columns} referencia a {referenced_table}.{referenced_columns}')

# Cierra la conexión a la base de datos
engine.dispose()
```
