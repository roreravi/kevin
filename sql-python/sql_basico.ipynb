Importación de las librerías necesarias:

```python
import pandas as pd
from sqlalchemy import create_engine
```

import pandas as pd: Importa la librería Pandas.

from sqlalchemy import create_engine: Importa la función create_engine de SQLAlchemy, que se utiliza para establecer la conexión con la base de datos.

Conexión a la Base de Datos SQLite con SQLAlchemy:

create_engine crea un objeto de "motor" que representa la conexión a la base de datos.

Conexión a la base de datos SQLite (asumiendo que el archivo de la base de datos se llama ecommerce.sqlite y está en el mismo directorio):

```python
# Creamos un motor para la base de datos SQLite
engine = create_engine('sqlite:///ecomerce.sqlite')

# Probamos la conexión (opcional)
with engine.connect() as connection:
    print("Conexión a la base de datos exitosa.")
```

```python
#Voy a listar las tablas de la base de datos
query = ''' 
            SELECT name 
            FROM sqlite_master 
            WHERE type = 'table' 
            ORDER BY name;
        '''
df_tablas = pd.read_sql_query(query, engine)
```

```python
df_tablas
```

create_engine('sqlite:///ecommerce.sqlite'): Crea un objeto engine que sabe cómo comunicarse con una base de datos SQLite ubicada en el archivo ecommerce.sqlite. La sintaxis 'sqlite:///' indica una base de datos SQLite local.

engine.connect(): Establece una conexión con la base de datos. Es importante cerrar la conexión después de usarla, y el with statement se encarga de esto automáticamente.

Ejecución de Consultas SQL Básicas con SQLAlchemy:

Introducción a las consultas SQL SELECT, FROM, WHERE, LIMIT.

Ejemplo 1: Seleccionar todas las columnas de la tabla customers y mostrar las primeras 5 filas usando Pandas:



```python
#INSPECCIONAMOS LAS COLUMNAS
# Ejecutamos una consulta SQL para seleccionar todos los clientes
query = ''' 
           SELECT * 
           FROM customers 
           LIMIT 5
        '''

# Leemos el resultado de la consulta directamente en un DataFrame de Pandas
df_customers = pd.read_sql(query, engine)

# Mostramos el DataFrame
print("Primeras 5 filas de la tabla 'customers':")
print(df_customers)
```

```python
df_customers.columns
```

```python
#PARA VER EL NUMERO DE FILAS
query = ''' 
            SELECT COUNT(1)
            FROM customers
        '''
pd.read_sql_query(query, engine)
```

query = "SELECT * FROM customers LIMIT 5": Define una cadena de texto que contiene la consulta SQL. SELECT * selecciona todas las columnas de la tabla customers, y LIMIT 5 restringe el resultado a las primeras 5 filas.

pd.read_sql(query, engine): Función de Pandas que ejecuta la consulta SQL (query) utilizando la conexión a la base de datos (engine) y carga el resultado en un DataFrame llamado df_customers.

Ejemplo 2: Seleccionar columnas específicas y filtrar por una condición (usando la tabla orders):

```python
#FILTROS
# Seleccionamos order_id, customer_id y order_status de la tabla orders
query_orders =  '''
                SELECT order_id, customer_id, order_status 
                FROM orders 
                WHERE order_status = 'delivered' 
                LIMIT 10
                '''
df_delivered_orders = pd.read_sql(query_orders, engine)
print("\nPrimeras 10 órdenes con estado 'delivered':")
print(df_delivered_orders)
```

WHERE order_status = 'delivered': Cláusula SQL que filtra las filas donde la columna order_status es igual a 'delivered'.

Ejemplo 3: Seleccionar datos de la tabla products y ordenar por nombre:

```python
query_products = '''
                    SELECT  product_id, 
                            product_category_name, 
                            product_name_lenght 
                    FROM products
                    WHERE product_category_name <> 'None'
                    ORDER BY product_name_lenght 
                    LIMIT 5
                ''' 
# Mostramos el DataFrame
print("Primeras 5 filas de la tabla 'customers':")
print(df_customers)
df_long_products = pd.read_sql(query_products, engine)
print("\nLos 5 productos con el nombre más largo:")
print(df_long_products)
```

ORDER BY product_name_lenght DESC: Cláusula SQL que ordena los resultados por la columna product_name_lenght en orden descendente (DESC).

Análisis Básico con Pandas de los Datos Extraídos:

Utilizar los DataFrames creados (df_customers, df_delivered_orders, df_long_products) para realizar análisis básicos como:

.head() y .info() para inspeccionar los DataFrames.
.value_counts() en columnas específicas para ver la distribución de valores.

.describe() para obtener estadísticas descriptivas (si las columnas son numéricas).

Filtrado básico en los DataFrames usando la sintaxis de Pandas.


```python
print("\nInformación del DataFrame de clientes:")
print(df_customers.info())

print("\nConteo de los diferentes estados de orden:")
query_order_status = "SELECT order_status FROM orders"
df_order_status = pd.read_sql(query_order_status, engine)
print(df_order_status['order_status'].value_counts())

print("\Estadísticas descriptivas de la longitud del nombre de los productos:")
query_product_length = "SELECT product_name_lenght FROM products"
df_product_length = pd.read_sql(query_product_length, engine)
print(df_product_length.describe())

# Filtrar clientes del estado de 'SP' usando Pandas
sp_customers = df_customers[df_customers['customer_state'] == 'SP']
print("\nPrimeros 5 clientes del estado de SP:")
print(sp_customers.head())
```
