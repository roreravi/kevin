30 Ejercicios de SQL: Consolidación de Fundamentos e Intermedios
Base de Datos: formula1.db (SQLite)
Herramienta: Python con SQLAlchemy y Pandas.
Instrucciones: Para cada ejercicio, escribe la consulta SQL necesaria para responder la pregunta y luego el código Python para ejecutarla y mostrar el resultado en un DataFrame de Pandas.

Nivel 1: Fundamentos (SELECT, FROM, WHERE, ORDER BY, LIMIT, DISTINCT)
Ejercicios de calentamiento para asegurar la comprensión de lo básico.

Listado Básico de Pilotos:

Pregunta: ¿Cuáles son los nombres y apellidos (forename, surname) de los primeros 10 pilotos registrados en la base de datos?
Circuitos en un País Específico:

Pregunta: ¿Cuáles son los nombres (name) y ubicaciones (location) de todos los circuitos que se encuentran en 'USA'?
Carreras en un Año Determinado:

Pregunta: ¿Cuáles son los nombres (name) de todas las carreras que se celebraron en el año 2023? (Si no hay en 2023, usa el año más reciente con datos, por ejemplo, 2022 o 2021).
Conteo Total de Constructores:

Pregunta: ¿Cuántos constructores diferentes hay registrados en la tabla constructors?
Temporadas Ordenadas (Descendente):

Pregunta: ¿Cuáles son los años (year) de todas las temporadas, ordenadas desde la más reciente hasta la más antigua? Muestra solo los primeros 5 años.
Pilotos Nacidos Después de una Fecha:

Pregunta: ¿Cuáles son los nombres (forename, surname) de los pilotos nacidos después del 1 de enero de 1995? Ordena por fecha de nacimiento ascendente.
Nacionalidades Únicas de Constructores:

Pregunta: ¿Cuáles son todas las nacionalidades únicas (nationality) de los constructores presentes en la base de datos?
Resultados con Puntos Específicos:

Pregunta: ¿Cuáles son los raceId, driverId y points para los resultados donde se obtuvieron exactamente 6 o 8 puntos? Muestra los primeros 10.
Fechas de Nacimiento Nulas:

Pregunta: ¿Hay algún piloto en la base de datos cuya fecha de nacimiento (dob) sea desconocida (es decir, NULL)? Si los hay, muestra sus driverRef y nombres.
Carreras con Nombres Similares:

Pregunta: ¿Cuáles son los nombres (name) de las carreras que incluyen la palabra "Grand" en su nombre? Muestra los primeros 5.
Nivel 2: Agregación y Agrupamiento (COUNT, SUM, AVG, MIN, MAX, GROUP BY, HAVING)
Ejercicios que requieren resumir y agrupar datos.

Puntos Totales por Piloto (Top 5):

Pregunta: ¿Cuál es la suma total de puntos (points) acumulados por cada piloto a lo largo de su carrera?
Muestra el driverId y la suma de puntos, ordenado de mayor a menor, para los 5 pilotos con más puntos. (Solo considera points > 0).
Número de Carreras por Año:

Pregunta: ¿Cuántas carreras se celebraron en cada año?
Muestra el año y el conteo de carreras, ordenado por año ascendente.
Promedio de Puntos por Constructor:

Pregunta: ¿Cuál es el promedio de puntos obtenidos por carrera para cada constructor que haya participado en la tabla constructor_results?
Muestra el constructorId y el promedio de puntos, ordenado de mayor a menor.
El Circuito Más Largo y el Más Corto:

Pregunta: ¿Cuál es la longitud (alt) del circuito con la mayor altitud y del circuito con la menor altitud?
Muestra solo los valores mínimo y máximo de alt.
Nacionalidades de Pilotos con Más de 10 Miembros:

Pregunta: ¿Cuáles son las nacionalidades de los pilotos que tienen 10 o más representantes en la base de datos?
Muestra la nacionalidad y el conteo de pilotos, ordenado de mayor a menor conteo.
Años con Menos de 15 Carreras:

Pregunta: ¿Cuáles son los años en los que se celebraron menos de 15 carreras?
Muestra el año y el número de carreras, ordenado por año.
Diferencia entre Mayor y Menor Puntuación en una Carrera:

Pregunta: ¿Cuál es la diferencia entre los puntos máximos y mínimos obtenidos en la carrera con raceId = 1?
Pilotos con una Cantidad Específica de Victorias (Posición 1):

Pregunta: ¿Cuáles driverId han logrado exactamente 5 victorias (position = 1) en sus carreras?
Muestra el driverId y el conteo de victorias.
Promedio de Puntos por Piloto en Carreras con Puntos:

Pregunta: ¿Cuál es el promedio de puntos por carrera para cada piloto, considerando solo las carreras donde obtuvieron puntos (points > 0)?
Muestra el driverId y el promedio, ordenado por promedio descendente.
Conteo de Resultados por Status (Solo Finalizados):

Pregunta: ¿Cuántos resultados tienen un statusId de 1 (que corresponde a 'Finished')?
Nivel 3: Uniones de Tablas (JOINs)
Ejercicios que combinan información de múltiples tablas.

Nombres de Carreras y Circuitos en 2022:

Pregunta: ¿Cuáles son los nombres de las carreras y los nombres de los circuitos donde se llevaron a cabo esas carreras en el año 2022?
Muestra el nombre de la carrera y el nombre del circuito.
Pilotos y sus Equipos en la Temporada 2021:

Pregunta: ¿Cuáles son los nombres completos de los pilotos y los nombres de los constructores (constructors.name) para los que corrieron en el año 2021?
Muestra el nombre del piloto y el nombre del constructor.
Victorias de Ferrari en 2020:

Pregunta: ¿Cuáles fueron las carreras (nombre de la carrera) en las que el constructor 'Ferrari' ganó (posición 1) en el año 2020?
Muestra el nombre de la carrera.
Circuitos que No Han Albergado Carreras desde 2020:

Pregunta: ¿Cuáles son los nombres (name) de los circuitos que no han sido utilizados para ninguna carrera desde el año 2020 (inclusive)?
Muestra el nombre del circuito y su ubicación. (Pista: LEFT JOIN y IS NULL o una subconsulta con NOT IN).
Pilotos con Múltiples Nacionalidades (Si Fuera el Caso - Ejercicio Concepto):

Pregunta: Si hubiera pilotos con la misma combinación de nombre y apellido, pero diferente nacionalidad (ej. por errores de datos), ¿cómo los identificarías?
(Pista: No esperes resultados reales con esta base de datos, es para practicar el GROUP BY en columnas unidas). ¿Cuántas nacionalidades distintas hay por cada forename + surname (nombre completo)?
Puntos Acumulados por Piloto y Constructor en una Carrera Específica:

Pregunta: Para la carrera con raceId = 1000 (o elige una existente, por ejemplo, raceId = 1052 para el Gran Premio de Arabia Saudita 2021), ¿cuántos puntos ganó cada piloto y para qué constructor?
Muestra el nombre del piloto, el nombre del constructor y los puntos obtenidos, ordenado por puntos descendente.
Carreras con 'Weather' en el Nombre del Circuito:

Pregunta: ¿Cuáles son los nombres de las carreras que se celebraron en circuitos cuyo nombre incluye la palabra "Weather"? (No hay un circuito "Weather", busca uno que tenga una 'e' y 'a' en el nombre o usa 'International' para un ejemplo más realista).
Muestra el nombre de la carrera y el nombre del circuito.
Estado Final de Resultados y Su Conteo Global:

Pregunta: ¿Cuántos resultados hay para cada estado final de carrera (status.status)?
Muestra el nombre del estado y el conteo de resultados, ordenado por el conteo descendente.
Pilotos y Sus Puntos Acumulados por Año:

Pregunta: ¿Cuántos puntos acumuló cada piloto en cada año de carrera?
Muestra el nombre del piloto, el año y la suma de puntos, ordenado por nombre de piloto y luego por año. (Pista: GROUP BY piloto y año).
Circuitos con Más de 30 Carreras Históricas:

Pregunta: ¿Cuáles son los nombres de los circuitos que han albergado más de 30 carreras en toda la historia de la Fórmula 1 registrada en la base de datos?
Muestra el nombre del circuito y el conteo de carreras, ordenado por conteo descendente.
