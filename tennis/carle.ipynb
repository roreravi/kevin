Jupyter Notebook: Análisis de Datos de Tenis de María Lourdes Carlé

1.  Importar Librerías

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#   Explicación:
#   -   pandas: Para la manipulación y análisis de datos tabulares.
#   -   numpy:   Para operaciones numéricas eficientes.
#   -   matplotlib.pyplot: Para la creación de visualizaciones básicas.
#   -   seaborn:   Para visualizaciones estadísticas más avanzadas y estéticas.
#   -   %matplotlib inline:  Para mostrar los gráficos directamente en el notebook.
```

2.  Cargar y Explorar los Datos

```python
#   Como los datos de Carlé están en un documento de Word, primero debemos
#   convertirlos a un formato estructurado (por ejemplo, CSV o diccionario de Python).
#   Para simplificar, voy a definir los datos directamente como diccionarios de Python.

#   Datos de Ranking
data_ranking = {
    "Año": [2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    "Ranking_Individual": [727, 865, 639, 440, 262, 161, 153, 95],
    "Ranking_Dobles": [None, 1080, 815, 644, 225, 175, 460, 124] #   'None' donde no hay dato
}

#   Datos de Victorias/Derrotas por Año
data_victorias_derrotas = {
    "Año": [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025],
    "Victorias": [7, 2, 0, 27, 9, 31, 33, 43, 51, 43, 48, 15],
    "Derrotas": [3, 3, 1, 7, 4, 12, 7, 27, 36, 30, 29, 14]
}

#   Convertir los diccionarios a DataFrames de pandas
df_ranking = pd.DataFrame(data_ranking)
df_victorias_derrotas = pd.DataFrame(data_victorias_derrotas)

#   Explorar los datos
print("Datos de Ranking:")
print(df_ranking.head())
print("\nDatos de Victorias/Derrotas:")
print(df_victorias_derrotas.head())

#   Explicación:
#   -   Los datos se organizan en diccionarios de Python, donde las claves son los nombres de las columnas,
#       y los valores son listas con los datos de cada columna.
#   -   `pd.DataFrame()` convierte los diccionarios en DataFrames, que son estructuras de datos tabulares
#       muy poderosas en pandas.
#   -   `df.head()` muestra las primeras filas del DataFrame para tener una idea de su contenido.
```

3.  Análisis del Ranking

```python
#   Visualizar la evolución del ranking individual
plt.figure(figsize=(10, 5))
plt.plot(df_ranking["Año"], df_ranking["Ranking_Individual"], marker='o')
plt.title("Evolución del Ranking Individual de Carlé")
plt.xlabel("Año")
plt.ylabel("Ranking Individual")
plt.gca().invert_yaxis()  #   Invertir el eje y para que los valores más bajos (mejor ranking) estén arriba
plt.grid(True)
plt.show()

#   Visualizar la evolución del ranking de dobles
plt.figure(figsize=(10, 5))
plt.plot(df_ranking["Año"], df_ranking["Ranking_Dobles"], marker='o')
plt.title("Evolución del Ranking de Dobles de Carlé")
plt.xlabel("Año")
plt.ylabel("Ranking de Dobles")
plt.gca().invert_yaxis()  #   Invertir el eje y
plt.grid(True)
plt.show()

#   Explicación:
#   -   `plt.figure(figsize=(10, 5))` crea una figura de 10x5 pulgadas.
#   -   `plt.plot()` genera un gráfico de líneas.
#   -   `plt.title()`, `plt.xlabel()`, `plt.ylabel()` establecen el título y las etiquetas de los ejes.
#   -   `plt.gca().invert_yaxis()` invierte el eje y. En el ranking de tenis, un número menor es mejor.
#   -   `plt.grid(True)` agrega una cuadrícula al gráfico.
#   -   `plt.show()` muestra el gráfico.
```

4.  Análisis de Victorias y Derrotas

```python
#   Calcular el porcentaje de victorias
df_victorias_derrotas['Porcentaje_Victorias'] = (df_victorias_derrotas['Victorias'] / (df_victorias_derrotas['Victorias'] + df_victorias_derrotas['Derrotas'])).fillna(0)

#   Visualizar Victorias y Derrotas por Año
plt.figure(figsize=(12, 6))
plt.bar(df_victorias_derrotas['Año'], df_victorias_derrotas['Victorias'], label='Victorias')
plt.bar(df_victorias_derrotas['Año'], df_victorias_derrotas['Derrotas'], label='Derrotas')
plt.xlabel('Año')
plt.ylabel('Número de Partidos')
plt.title('Victorias y Derrotas por Año')
plt.legend()
plt.show()

#   Visualizar el porcentaje de victorias
plt.figure(figsize=(12, 6))
plt.plot(df_victorias_derrotas['Año'], df_victorias_derrotas['Porcentaje_Victorias'], marker='o', color='green')
plt.xlabel('Año')
plt.ylabel('Porcentaje de Victorias')
plt.title('Porcentaje de Victorias por Año')
plt.grid(True)
plt.show()


#   Explicación:
#   -   Se calcula el porcentaje de victorias y se agrega como una nueva columna al DataFrame.
#   -   `plt.bar()` crea un gráfico de barras para comparar victorias y derrotas por año.
#   -   `plt.legend()` muestra la leyenda para identificar las barras de victorias y derrotas.
#    -   `fillna(0)`:  Remplaza los valores NaN (Not a Number) con 0. Esto es importante porque si en algún año
#        no hay ni victorias ni derrotas, la división resultaría en NaN.
```

5.  Preparación de Datos de Servicio (Ejemplo Simplificado)

```python
#   Para este ejemplo, vamos a crear datos ficticios de servicio.
#   En un escenario real, estos datos vendrían de un archivo CSV o una base de datos.
#   Este es el tipo de datos que necesitaríamos para replicar el análisis del PDF.

data_servicio = {
    "Lado": ["Deuce", "Deuce", "Ad", "Ad"],
    "Direccion": ["Ancho", "T", "Ancho", "Cuerpo"],
    "Porcentaje": [0.39, 0.48, 0.52, 0.35],
    "Puntos_Ganados": [0.75, 0.63, 0.64, 0.62],
    "Velocidad_Promedio": [170, 175, 173, 167],  #   km/h
    "Tipo_de_Efecto": ["Plano", "Slice", "Plano", "Slice"]
}
df_servicio = pd.DataFrame(data_servicio)

print("\nDatos de Servicio (Ejemplo):")
print(df_servicio.head())
```

6.  Análisis del Servicio (Ejemplo Simplificado)

```python
#   Efectividad del servicio por dirección
plt.figure(figsize=(8, 5))
sns.barplot(x='Direccion', y='Puntos_Ganados', hue='Lado', data=df_servicio)
plt.title('Efectividad del Primer Servicio por Dirección y Lado')
plt.ylabel('Porcentaje de Puntos Ganados')
plt.show()

#   Velocidad del servicio por tipo de efecto
plt.figure(figsize=(8, 5))
sns.barplot(x='Tipo_de_Efecto', y='Velocidad_Promedio', hue='Lado', data=df_servicio)
plt.title('Velocidad del Primer Servicio por Tipo de Efecto y Lado')
plt.ylabel('Velocidad Promedio (km/h)')
plt.show()


#   Explicación:
#   -   `sns.barplot()` crea gráficos de barras para comparar la efectividad y velocidad del servicio
#       en función de la dirección, el tipo de efecto y el lado de la cancha.
#   -   `hue`  se utiliza para diferenciar los datos por el lado de la cancha (Deuce o Ad).
```
